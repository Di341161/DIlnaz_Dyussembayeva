```python
class Users:
    def __init__(self, name, surename, birthdate, email):
        self.name=name
        self.surename=surename
        self.birthdate=birthdate
        self.email=email
    def getAge(self):
        return 2023-self.birthdate
```

```python
my_data= Users("Felix", "Lee", 1988, "felix.lee@gmail.com")
print(my_data.name)
print(my_data.surename)
print(my_data.birthdate)
print(my_data.email)
print(my_data.getAge())
```

```python
users = []
with open('users.txt', 'r') as file:
    for line in file:
        parts = line.strip().split(',')
        user = Users(parts[1], parts[2], int(parts[4]), parts[3])
        users.append(user)
```

```python
for user in users:
    print(user.name, user.surename, user.email, user.getAge())
```

```python
def filter(low, high, users=[]):
    for user in users:
        age=user.getAge()
        if age> low and age<high:
            print(user.email, ' ', user.getAge())  
```

```python
filter(30,50,users)
```
